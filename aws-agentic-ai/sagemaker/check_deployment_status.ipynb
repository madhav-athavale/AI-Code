```python
import boto3

sm_client = boto3.client('sagemaker', region_name='us-east-1')

response = sm_client.describe_endpoint(
    EndpointName='xgboost-synthetic-classifier'
)

print(f"Status:           {response['EndpointStatus']}")
print(f"Secondary status: {response['FailureReason'] if response['EndpointStatus'] == 'Failed' else 'N/A'}")
```
