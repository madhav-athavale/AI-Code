```python
import csv
import chromadb
from chromadb.utils import embedding_functions 
```

```python
documents = []
metadatas = []
ids = []
id = 1
```

```python
with open('menu_items.csv') as file:
    lines = csv.reader(file)
    for i, line in enumerate(lines):
        if i == 0:
            continue
        documents.append(line[1])
        ids.append(str(id))
        metadatas.append({"item_id": line[0]})
        id+=1
    
```

```python
chroma_client = chromadb.PersistentClient(path='my_vectordb')
```

```python
sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name='all-mpnet-base-v2')
```

```python
collection = chroma_client.get_or_create_collection(name="my_collection", embedding_function=sentence_transformer_ef)
```

```python
import time
start = time.time()
collection.add(documents=documents,ids=ids)
end= time.time()

print((end - start) )
```

```python
results = collection.query(query_texts = ["donut"], n_results = 5, include=['documents','distances'])
```
