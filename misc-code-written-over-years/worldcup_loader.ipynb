```python
import json
import streamlit
from  fastapi import FastAPI
```

```python
with open("worldcup.groups.json") as fp:
    str = fp.read()
```

```python
js = json.loads(str)
```

```python
print(js)
```

```python
print(js['groups']['name'=='Group A']['teams'][2])
```

```python
for gr in js['groups']:
    print(gr['name'], gr['teams'][2])
```

```python
app = FastAPI()
```

```python
@app.get("/group/{id}")
def read_item(id: int):
    # item_id is validated as an integer automatically
    return {"Group": js['groups']['name'==id]}
```
