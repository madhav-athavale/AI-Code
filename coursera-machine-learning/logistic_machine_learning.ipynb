Logistic machine learning -Breast cancer dataset

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

cancer = load_breast_cancer()

print(cancer.DESCR) # Print the data set description
```

```python
df = pd.DataFrame(data=cancer.data, columns =cancer.feature_names)
```

```python
y = cancer.target


    
```

```python
class_labels = list(cancer.target_names)
```

```python
df.shape
```

```python
y=cancer.target
y
```

```python
X = df.loc[:,df.columns[0:30]]
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 143, train_size= 426,random_state=0)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression(max_iter=1000, random_state=10)
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled)[:,1]
```

```python
print(f"Overall Model Accuracy: {accuracy_score(y_test, y_pred):.4f}\n")
#"Confusion Matrix
print (confusion_matrix(y_test,y_pred))
#Classifiication Report
print(classification_report(y_test,y_pred,target_names=class_labels))
```
